```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Baca gambar
image = cv2.imread('img_gray.jpg')

# Konversi gambar ke HSV
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Fungsi bantu untuk menampilkan hasil deteksi warna
def tampilkan_citra(mask, warna):
    hasil = cv2.bitwise_and(image, image, mask=mask)
    plt.imshow(cv2.cvtColor(hasil, cv2.COLOR_BGR2RGB))
    plt.title(warna)
    plt.axis("off")
    plt.show()
    return hasil

# ===================== DETEKSI WARNA MERAH =====================
# Warna merah butuh dua rentang HSV
lower_red1 = np.array([0, 120, 70])
upper_red1 = np.array([10, 255, 255])
lower_red2 = np.array([170, 120, 70])
upper_red2 = np.array([180, 255, 255])
mask_red = cv2.inRange(hsv, lower_red1, upper_red1) | cv2.inRange(hsv, lower_red2, upper_red2)
hasil_merah = tampilkan_citra(mask_red, "MERAH")

# ===================== DETEKSI WARNA HIJAU =====================
lower_green = np.array([35, 100, 100])
upper_green = np.array([85, 255, 255])
mask_green = cv2.inRange(hsv, lower_green, upper_green)
hasil_hijau = tampilkan_citra(mask_green, "HIJAU")

# ===================== MENAMPILKAN HASIL GABUNGAN =====================
# Gabungkan hasil merah dan hijau
mask_combined = cv2.bitwise_or(mask_red, mask_green)
hasil_combined = cv2.bitwise_and(image, image, mask=mask_combined)

# Menampilkan hasil gambar dengan warna merah dan hijau yang lebih jelas
plt.imshow(cv2.cvtColor(hasil_combined, cv2.COLOR_BGR2RGB))
plt.title("Merah dan Hijau")
plt.axis("off")
plt.show()
```

```python
def tampilkan_dengan_kotak(mask, warna_label, warna_bgr):
    hasil = image.copy()
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for i, contour in enumerate(contours):
        if cv2.contourArea(contour) > 500:  # batas minimum area
            x, y, w, h = cv2.boundingRect(contour)
            cv2.rectangle(hasil, (x, y), (x + w, y + h), warna_bgr, 2)
            cv2.putText(hasil, f"{warna_label} {i+1}", (x, y - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, warna_bgr, 2)
    
    plt.imshow(cv2.cvtColor(hasil, cv2.COLOR_BGR2RGB))
    plt.title(f'Deteksi Warna: {warna_label}')
    plt.axis('off')
    plt.show()
```

```python
# Fungsi untuk menampilkan histogram dari gambar
def tampilkan_histogram(gambar, judul):
    color = ('b', 'g', 'r')  # Warna channel BGR
    plt.figure()
    plt.title(f"Histogram - {judul}")
    plt.xlabel("Intensitas Pixel")
    plt.ylabel("Jumlah Pixel")
    
    for i, col in enumerate(color):
        hist = cv2.calcHist([gambar], [i], None, [256], [0, 256])
        plt.plot(hist, color=col)
        plt.xlim([0, 256])
    plt.show()

# Histogram citra kontras (gambar asli)
tampilkan_histogram(image, "CITRA KONTRAS")

# Histogram hasil deteksi
tampilkan_histogram(hasil_merah, "MERAH")
tampilkan_histogram(hasil_hijau, "HIJAU")
tampilkan_histogram(hasil_biru, "BIRU")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Baca gambar (ganti 'backlight.jpg' dengan nama file gambar Anda)
img = cv2.imread("PaceCode1.jpg")

# 1. Gambar Asli
original = img.copy()

# 2. Konversi ke grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 3. Cerahkan gambar grayscale (dengan menaikkan semua pixel)
bright_gray = cv2.convertScaleAbs(gray, alpha=1, beta=40)  # beta = brightness

# 4. Perkuat kontras dengan CLAHE
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
contrast_gray = clahe.apply(gray)

# 5. Cerahkan + Perkuat kontras
bright_contrast_gray = clahe.apply(bright_gray)

# -------------------- menampilkan beberapa warna pada Gambar --------------------
titles = ["1. Gambar Asli", 
          "2. Gambar Gray", 
          "3. Gray + Cerah", 
          "4. Gray + Kontras", 
          "5. Gray + Cerah + Kontras"]

images = [original, gray, bright_gray, contrast_gray, bright_contrast_gray]

plt.figure(figsize=(15, 8))
for i in range(5):
    plt.subplot(2, 3, i+1)
    if len(images[i].shape) == 2:
        plt.imshow(images[i], cmap='gray')
    else:
        plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```
